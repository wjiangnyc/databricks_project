```sql
MERGE INTO retail_uc.facts.web_returns AS t
USING (
  SELECT
    wr_order_number,
    wr_item_sk,
    wr_returned_date_sk,
    wr_reason_sk,
    wr_web_page_sk,
    wr_return_quantity,
    wr_return_amt,
    wr_return_tax,
    wr_return_amt_inc_tax,
    wr_fee,
    wr_return_ship_cost,
    wr_refunded_cash,
    wr_reversed_charge,
    wr_store_credit,
    wr_net_loss,
    d_date,
    d_year,
    d_moy,
    d_dom,
    d_day_name
  FROM postgres_catalog.retail.web_returns
) s
ON  t.wr_order_number = s.wr_order_number
AND t.wr_item_sk      = s.wr_item_sk
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *;
```
