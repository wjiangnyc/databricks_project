```sql
CREATE OR REPLACE TABLE retail_uc.analytics.returns_by_category AS
SELECT
  i.i_category AS category,
  d.d_year,
  d.d_moy AS month,
  SUM(wr_return_quantity) AS total_returns,
  SUM(wr_net_loss) AS total_net_loss
FROM retail_uc.facts.web_returns wr
JOIN retail_uc.dims.item i ON wr.wr_item_sk = i.i_item_sk
JOIN retail_uc.dims.date_dim d ON wr.wr_returned_date_sk = d.d_date_sk
GROUP BY i.i_category, d.d_year, d.d_moy;
```
