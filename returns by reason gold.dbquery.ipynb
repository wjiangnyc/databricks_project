```sql
CREATE OR REPLACE TABLE retail_uc.analytics.returns_by_reason AS
SELECT
  r.r_reason_desc AS return_reason,
  SUM(wr_return_quantity) AS total_returns,
  SUM(wr_net_loss) AS total_net_loss
FROM retail_uc.facts.web_returns wr
JOIN retail_uc.dims.reason r ON wr.wr_reason_sk = r.r_reason_sk
GROUP BY r.r_reason_desc;
```
